MATHEMATICAL CONVENTIONS
r_t = mu + epsilon_t; epsilon_t = sqrt(h_t) * z_t, elementwise.
GARCH: h_next = omega + alpha * epsilon_t**2 + beta * h_t.
GJR: h_next additionally includes gamma * epsilon_t**2 * I(epsilon_t < 0).
With symmetric standard Gaussian innovations, p_GARCH = alpha + beta;
p_GJR = alpha + beta + gamma/2; long-run variance = omega / (1 - p).
Initialize the observed history at long-run variance; initialize futures at the next
variance calculated from the final observed residual and its associated variance.
Annual asset return: X[m,i] = product_t(1 + r[m,t,i]) - 1.
Buy-and-hold annual portfolio return: X @ weights; annual loss: -X @ weights.
Annual fitted mean and covariance are calculated from the same X_opt for all strategies.
Exact ERC: weights_i * (Sigma @ weights)_i / (weights.T @ Sigma @ weights) = 1/A.
With positive-definite covariance, positive budgets, and standard long-only unit budget,
the normalized exact ERC solution is unique. Its mean is the maximum over that
singleton feasible ERC set; no separate return/risk trade-off is introduced.
r_star = mu_annual_hat @ weights_erc.
Markowitz: minimize annual variance subject to mu_annual_hat @ weights = r_star.
CVaR: minimize annual scenario tail loss subject to the same equality.
Equal fitted means do not imply equal realized or oracle expected returns.
Compounded annual portfolios are not covered by the daily Gaussian VaR/CVaR formula.
